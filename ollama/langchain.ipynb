#### LCEL 파이프라인
- LCEL(LangChain Expression Language): | 연산자로 컴포넌트를 연결하는 파이프라인 문법 / 랭체인에서 사용하는 표현 문법
  - ex) prompt | llm | parser => 왼쪽에서 오른쪽으로다음 절차를 밟어 나감
  - RunnableSequence 생성
    - RunnablePassthrough: 입력을 그대로 통과시켜 다른 키와 병합할 때 사용
    - RunnableLambda: 체인을 일반 파이썬 함수와 연결
    - RunnableParallel: 병렬처리. 여러 체인을 동시에 실행하고 결과를 dict 병합할 때 사용
    - .assign(): 체인 중간에 새로운 키 추가
- invoke(): 단일 동기 호출. 답변이 올 때 까지 기다렸다가 한방에 보여줌]stream(), 스트리밍, batch() ->배치 -질문 여러 개 할 수 있다

```python
prompt = ChatPromptTemplate.from_messages([
  ("system","당신은 {role}"),
  ("human","{question}")
])

chain = prompt | watson_llm

response = chain.invoke({
  "role":"보안",
  "question": "xxs란?"
})

print(response)
print(response.content)
```

```python
# gt : >
# ge: >=
# lt : <
# le: <=

class User(BaseModel):
  name: str
  age: int = Field(gt=0)

# 객체 생성
user = User(name="홍길동", age="20")
print(user)
print(type(user.age))
```

```python
# 감정, 점수, 이유 키워드 3개
# Pydantic 모델 정의
class SentimentResult(BaseModel):
  sentiment:Literal['positive','negative','neutral']
  score:float = Field(ge=0.0, le=1.0, description="감정 강도")
  reason: str = Field(description="판단 근거 한 문장")
  keywords:list[str] = Field(description="핵심 키워드 3개 이내")

# 파서 정의
pydantic_parser = PydanticOutputParser(pydantic_object = SentimentResult)

# prompt
# {format_instructions}: pydantic_parser의 필드를 참고해서 작성해줘
prompt = ChatPromptTemplate.from_messages([
  ('system', '김정 분석 전문가입니다.\n\n {format_instructions}'),
  ('human', '다음 텍스트의 감정을 분석하세요: {text}'),
]).partial(format_instructions=pydantic_parser.get_format_instructions())

chain = prompt |  exaone_llm | pydantic_parser
result = chain.invoke({
  "text":"오늘은 최악의 하루였습니다. 모든 게 잘못됐어요."
})

print(type(result))
print(result)
print(result.sentiment)
print(result.score)
print(result.reason)
print(result.keywords)
```

```python
# 여러 개 넣기

texts = [
  "이 제품 정말 별로예요. 돈 낭비였습니다.",
  "그냥 평범한 하루였어요. 특별한 건 없었어요.",
  "와, 생각보다 훨씬 맛있어요! 강력 추천합니다",
  "서비스가 너무 불친절하고 음식도 늦게 나왔어요.ㄴ"
]


chain = prompt |  watson_llm | pydantic_parser
results = chain.batch([{'text': t} for t in texts])

print("=====감정 분석 결과=====")
for text,r in zip(texts,results):
  bar = ' ' * int(r.score * 10) + ' ' * ( 10 - int(r.score * 10))
  emoji = {
    'positive': '😁',
    'negative':'🤬',
    'neutral':'🙂'
  }[r.sentiment]
  print(f'{emoji}[{r.sentiment:8s}] {bar} {r.score:.2f}')
  print(f' 입력: {text[:30]}...')
  print(f' 이유: {r.reason}')

```

```python
class AdvancedSentiment(BaseModel):
  sentiment:Literal['positive','negative','neutral']
  score:float = Field(ge=0.0, le=1.0, description="감정 강도")
  category: Literal['product','service','food','general'] = Field(description="리뷰 카테고리"), 
  summary:str = Field(description="한 줄 요약")
  action:str = Field(description="비지니스 조치 방향")

# 파서 정의
pydantic_parser = PydanticOutputParser(pydantic_object = AdvancedSentiment)

# prompt
# {format_instructions}: pydantic_parser의 필드를 참고해서 작성해줘
prompt = ChatPromptTemplate.from_messages([
  ('system', '당신은 CX(고객경험) 분석가입니다.\n\n {format_instructions}'),
  ('human', '{text}'),
]).partial(format_instructions=pydantic_parser.get_format_instructions())

chain = prompt |  watson_llm | pydantic_parser
result = chain.invoke({
  "text":"배송은 빠른데 포장이 너무 허술해요"
})


print(result)
print(result.sentiment)
print(result.score)
print(result.category)
print(result.summary)
print(result.action)
```

#### RunnableLambda
- chain => RunnableSequence 생성하기 떄문에 일반 파이썬 함수는 바로 연결 불가
- 일반 함수를 감싸는 용도로 사용

```python
def upper(text):
  return text.upper()

prompt = ChatPromptTemplate.from_messages([
  ("system", "당신은 언어 전문가입니다."),
  ("human", "{topic}에 대해 설명해줘")
])
model = watson_llm
parser = StrOutputParser()
runnable = RunnableLambda(upper)

chain = prompt | model | parser | runnable
result = chain.invoke({"topic":"python"})
print(result)
```

```python
# 후처리 함수
def remove_newline(text):
  return text.replace("\n","🍉")

prompt = ChatPromptTemplate.from_template("{question}에 대해 답변해주세요")
model = watson_llm
parser = StrOutputParser()
runnable = RunnableLambda(remove_newline)
chain = prompt | model | parser | runnable
result = chain.invoke({"question": "파이썬의 리스트"})
print(result)
```

#### RunnablePassThrough
변수에 들어온 값이 그대로 전달된다.

```python
prompt = ChatPromptTemplate.from_template("{question}에 대해 답변해줘")
model = watson_llm
parser = StrOutputParser()
runnable = RunnableLambda(lambda x:f"Q: {x['question']}\nA: {x['answer']}")

chain = (
  {
  "question": RunnablePassthrough(), 
  "answer": prompt | model | parser 
  } | runnable
)

result = chain.invoke("파이썬이란?")
print(result)
```

#### RunnableParallel
- 여러 chain을 호출할 때 사용
- 실제 병렬처리 성능은 컴퓨터 자원에 영향을 받는다 (cpu, gpu..)

```python
model = watson_llm
parser = StrOutputParser()
summary_prompt = ChatPromptTemplate.from_template("다음 텍스트를 한 문장으로 요약해줘: {text}")
keyword_prompt = ChatPromptTemplate.from_template("다음 텍스트의 핵심 키워드 3개를 콤마로 나열해줘: {text}")

# 요약
summary_chain = summary_prompt | model | parser

# 키워드
keyword_chain = keyword_prompt | model | parser

# 한꺼번에 병렬로 처리
parallel_chain = RunnableParallel(summary=summary_chain, keyword=keyword_chain)
result = parallel_chain.invoke({"text":"인공지능은 인간의 지능을 모방하는 기술로..."})
print(result["summary"])
print(result["keyword"])
```

#### .assign(): dict를 반환하는 결과에 붙여서 사용 가능

```python
analysis_prompt = ChatPromptTemplate.from_template("다음을 리뷰해줘 {review}")
summary_prompt = ChatPromptTemplate.from_template("한줄 요약해줘 {review}")
sentiment_prompt = ChatPromptTemplate.from_template("감정분석좀 해줘 긍정/부정/중립 중 하나로 해줘 {review}")
model = watson_llm
parser = StrOutputParser()
res = {"text": RunnablePassthrough()}


# res(llm 답변)에 들어간다
analysis_chain = analysis_prompt | model |  parser | res
full_chain = (RunnablePassthrough
              .assign(summary = summary_prompt | model | parser)
              .assign(sentiment = sentiment_prompt | model | parser)
              .assign(analysis=analysis_chain))
result = full_chain.invoke({"review":"정말 맛있는 음식이였어요!"})
print(result)
```

#### 멀티턴 대화
- LLM: stateless (이전의 대화 내용을 기억 못함)
  - 대화 이력 관리를 위해 messages 리스트에 누적 필요

```python
messages = [
  SystemMessage(content="당신은 친절한 파이썬 튜터입니다. 한국어로 대답하세요")
]

def chat(user_input):
  messages.append(HumanMessage(user_input))
  response = watson_llm.invoke(messages)
  content = response.content
  messages.append(AIMessage(content))
  return content

print(chat("파이썬이란?"))
print(chat("방금말한 내용의 장점 3가지는?"))
print(chat("그 중 첫번째 장점에 대해 예시 코드 작성해줘"))
print(chat("두 번째 장점에 대해 예시 코드 작성해줘"))

for m in messages:
  print(f"[{m.type}] {m.content[:40]}..." )
```

#### RunnableWithMessageHistory(메시지 자동 이력 관리)
- 세션별로 대화 이력을 관리해야 함
- InMemoryChatMessageHistory() : 메모리를 이용해서 저장
  - 메모리에 대화 저장 후 대화 기록 관리

메모리에 쌓는다면 계속 길어진다.
처리할 수 있는 context 용량이 넘어가면 context overflow가 생긴다.

```python
history = InMemoryChatMessageHistory()

def chat(user_input):
  history.add_user_message(user_input)
  response = watson_llm.invoke(history.messages) # 전체 대화 이력 같이 보내기
  content = response.content
  history.add_ai_message(content)
  return content

# context: 대화의 맥락을 알게 해주는 컨텐츠들
print(chat("파이썬이란?"))
print(chat("방금말한 내용의 장점 3가지는?"))
print(chat("그 중 첫번째 장점에 대해 예시 코드 작성해줘"))
print(chat("두 번째 장점에 대해 예시 코드 작성해줘"))

for m in history.messages:
  print(f"[{m.type}] {m.content[:40]}..." )
```

```python
# 여러 사람과 대화하는 부분 저장 -> 세션별로 저장

# 세션별 이력 저장소
store = {}

def get_session_history(session_id) -> BaseChatMessageHistory:
  if session_id not in store:
    store[session_id] = InMemoryChatMessageHistory()
  return store[session_id]

system_prompt="당신은 {role}입니다. 한국어로 답변하세요"

# 프롬프트 생성
# from_template 유저 메세지 하나만
# from_messages 여러개 보낼 때
prompt = ChatPromptTemplate.from_messages([
  ("system", system_prompt),
  MessagesPlaceholder(variable_name="history"), # 대화이력 삽입될 위치
  ("human", "input")
])
model = watson_llm
parser = StrOutputParser()

# LLM 생성
# 체인 생성
chain = prompt | model | parser

# 1. session_id 이력 조회
# 2. history 키에 이력 삽입 -> prompt -> llm실행 -> 대화 나눈 이력을 history에 추가
with_history = RunnableWithMessageHistory(chain, get_session_history, input_messages_key="input", history_messages_key="history")

# session 생성
cfg_a={"configurable":{"session_id":"user_alice"}}
cfg_b={"configurable":{"session_id":"user_bob"}}

# Alice 대화
res1 = with_history.invoke({"role":"파이썬 튜터", "input":"안녕"}, config=cfg_a)
res2 = with_history.invoke({"role":"파이썬 튜터", "input":"내 이름은 Alice"}, config=cfg_a)
res4 = with_history.invoke({"role":"파이썬 튜터", "input":"내 이름 뭐야?"}, config=cfg_a)

# bob 대화
res3 = with_history.invoke({"role":"파이썬 튜터", "input":"안녕"}, config=cfg_b)
res4 = with_history.invoke({"role":"파이썬 튜터", "input":"파이썬 변수 타입은?"}, config=cfg_b)

print("Alice", res2)
print("Alice", res4)
```

- 대화가 길어지면 이력 전체를 LLM에 넘기는 경우 Context Window를 초과
- Memory에 어떤 정보를 얼마나 보존할 지 결정
- 전략
  - 최근 K 턴만 유지
  - 오래된 대화 요약
  - 요약 + 최근 대화

```python
# 최근 K턴만 유지

# 세션별 이력 저장소
store = {}

def get_session_history(session_id) -> BaseChatMessageHistory:
  if session_id not in store:
    store[session_id] = InMemoryChatMessageHistory()

  # 최근 몇 개만 유지
  history = store[session_id]

  if len(history.messages) > 8:
    history.messages[:] = history.messages[-8:]
  
  return history

model = watson_llm
parser = StrOutputParser()
chain = prompt | model | parser
with_history = RunnableWithMessageHistory(
  chain, get_session_history, input_messages_key="input", history_messages_key="history"
)

for i in range(10):
  with_history.invoke({"role":"파이썬 튜터", "input": f"{i}번째 질문"}, config=cfg_a)

history = get_session_history("user_alice")
for msg in history.messages:
  print(msg)

print(len(history.messages))
```

```python
# 오래된 대화 요약

# history 요약용 체인 생성
summary_prompt = ChatPromptTemplate.from_messages([
  ("system", "다음 대화 내용을 핵심만 3문장 이내로 한국어로 요약해주세요"),
  ("human", "{conversation}")
])

summary_chain = summary_prompt | model | parser

# 요약 기능이 포함된 클래스
class SummarizedChatHistory(InMemoryChatMessageHistory):
  """대화가 max_turns 초과 시 오래된 메세지 요약하기"""
  max_turns:int = Field(default=6)
  summary:str = Field(default="")

  def _maybe_summarize(self):
    # 언제 요약할건지
    if len(self.messages) > self.max_turns * 2:
      # 오래된 내용 찾기
      cutoff = len(self.messages) // 2
      old_msgs = self.messages[:cutoff]
      new_msgs = self.messages[cutoff:]

      # 요약할 텍스트 구성
      conv_text = "\n".join(f"{'사용자' if isinstance(m,HumanMessage) else 'AI'}:{m.content}" for m in old_msgs)

      # 기존 요약이 있으면 함께 포함
      if self.summary:
        conv_text = f"[이전 요약]\n{self.summary}\n\n[새대화]\n{conv_text}"
      
      # 요약 실행
      self.summary = summary_chain.invoke({'conversation': conv_text})

      # 오래된 메세지 제거 후 요약본 SystemMessage로 앞에 삽입
      self.messages.clear()
      self.messages.append(SystemMessage(content=f"[이전 대화 요약]\n{self.summary}"))
      self.messages.extend(new_msgs)

  def add_message(self, message):
    super().add_message(message)
    self._maybe_summarize()

store = {}

def get_session_history(session_id) -> BaseChatMessageHistory:
  if session_id not in store:
    store[session_id] = SummarizedChatHistory(max_turn=6)
  return store[session_id]

# 프롬프트 생성
system_prompt="당신은 친절한 AI 어시스턴트입니다. 한국어로 답변하세요"
prompt = ChatPromptTemplate.from_messages([
  ("system", system_prompt),
  MessagesPlaceholder(variable_name="history"), # 대화이력 삽입될 위치
  ("human", "input")
])
model = watson_llm
parser = StrOutputParser()
chain = prompt | model | parser

with_history = RunnableWithMessageHistory(chain, get_session_history, history_messages_key="history")

cfg_a={"configurable":{"session_id":"user_alice"}}

# 임의 대화 생성
questions = [
  "파이썬이란?",
  "방금 설명한 파이썬의 장점은?",
  "단점은 뭐야?",
  "어떤 분야에 많이 쓰여?",
  "입문자에게 추천하는 학습 순서는?",
  "좋은 파이썬 책 추천해줘",
  "무료로 배울 수 있는 사이트 알려줘",
  "이전에 내가 뭘 물어봤는지 기억나?"
]

for i, q in enumerate(questions, 1):
  response = with_history.invoke({"input":q}, config=cfg_a)
  print(f"\n{i}턴 q:{q}")
  print(f"\n      a:{response[:60]}...")

  # 요약 여부 확인
  history = get_session_history("user_alice")
  if history.summary:
    print(f"/n 요약 처리! 현재 메시지 수 {len(history.messages)}")
    print(f"\n 요약 내용 {history.summary[:80]}...")
```
