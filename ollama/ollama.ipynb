```python

from pydantic import BaseModel, Field 

class SentimentResult(BaseModel):
  sentiment: str
  score: float = Field(ge=0.0, le=1.0)
  # keywords: list[str]

prompt="""
다음 문장 감정 분석
단 format으로 주어진 구조로 출력

문장: 오늘 정말 행복한 하루였다
"""


response = chat(
    model='qwen2.5:7b',
    messages=[{"role": "user","content": prompt}],
    format= SentimentResult.model_json_schema(), # 이 스키마를 참고하여 문자열을 생성
    stream=False
)


# LLM이 생성한 텍스트는 단순히 '문자열'
# 이를 실제 파이썬 객체로 바꿔주고, 동시에 score 필드에 설정하신 Field(ge=0.0, le=1.0) 제약 조건이 잘 지켜졌는지 검증
result = SentimentResult.model_validate_json(response.message.content)
print(result)
print(result.sentiment)
print(result.score)
```

### tool calling
- 정의된 함수를 호출

```python
def calculator(a,b):
  return a+b
  
```

```python
# 스키마 정의
tools = [
    {
        "type": "function",
        "function": {
            "name": "calculator",
            "description": "두 숫자를 더한다.",
            "parameters": {
                "type": "object",
                "properties": {
                    "a": {
                        "type": "integer",
                        "description": "첫 번째 숫자"
                    },
                    "b": {
                        "type": "integer",
                        "description": "두 번째 숫자"
                    }
                },
                "required": ["a","b"]
            }
        }
    }
]
```

```python
# 요청을 받아보는 상태
response = chat(
    model='qwen2.5:7b',
    messages=[{"role": "user", "content": "안녕 오늘 날씨가 너무 좋다. 25와 36을 더해줘"}],
    tools = tools
)

# role='assistant' content='' thinking=None images=None tool_name=None tool_calls=[ToolCall(function=Function(name='calculator', arguments={'a': 25, 'b': 36}))]
print(response.message)
```

```python
# llm이 필요하다면 tool을 실행
tool_call = response.message.tool_calls[0]
args = tool_call.function.arguments

result = calculator(args['a'], args['b'])

# tool 실행 결과를 llm으로 보내기
messages = [
  {"role":"user", "content":"25와 36을 더해줘"},
  response.message,
  {"role":"tool", "content":str(result)},
]

final_response = chat(
    model='qwen2.5:7b',
    messages=messages
)

print(final_response.message.content)
```

```python
import json

# ① 실제로 실행될 함수들
def get_weather(city: str) -> dict:
    # 실제로는 날씨 API 호출 — 여기선 더미 데이터
    data = {
        "서울": {"temp": 22, "condition": "맑음"},
        "부산": {"temp": 25, "condition": "흐림"},
    }
    return data.get(city, {"temp": 0, "condition": "알 수 없음"})

# ② LLM에게 "이런 Tool이 있다"고 알려주는 스키마
tools = [
    {
        "type": "function",
        "function": {
            "name": "get_weather",
            "description": "특정 도시의 현재 날씨를 조회한다",
            "parameters": {
                "type": "object",
                "properties": {
                    "city": {
                        "type": "string",
                        "description": "날씨를 조회할 도시 이름 (예: 서울, 부산)"
                    }
                },
                "required": ["city"]
            }
        }
    }
]

messages = [{"role": "user", "content": "서울 날씨 알려줘"}]

# ③ LLM 호출 — Tool 목록 전달
response = chat(
    model="qwen2.5:7b",
    messages=messages,
    tools=tools,
)

print(response.message)


# ④ LLM이 Tool 호출을 요청했으면 실제로 실행
if response.message.tool_calls:
    for tool_call in response.message.tool_calls:
        name = tool_call.function.name
        args = tool_call.function.arguments

        # 함수 이름으로 실제 함수 매핑
        tool_map = {"get_weather": get_weather}
        result = tool_map[name](**args)
        print(f"Tool 실행: {name}({args}) → {result}")

        # ⑤ Tool 결과를 대화 이력에 추가
        messages.append(response.message)          # LLM의 tool_calls 메시지
        messages.append({
            "role": "tool",
            "content": json.dumps(result, ensure_ascii=False),
        })

# ⑥ 결과를 포함해서 LLM에 다시 전달 → 최종 자연어 응답
final = chat(model="qwen2.5:7b", messages=messages)
print(final.message.content)
# "서울의 현재 날씨는 22°C이며 맑습니다."


```

```python
# Tool 함수들
def get_weather(city):
    return {"temp": 12, "condition": "비"}

def need_umbrella(condition):
    return {"result": condition in ["비", "눈"]}

def calculate(expression):
    return {"result": eval(expression)}  # 실제론 안전한 파서 사용

# Tool 스키마 목록
tools = [
    {"type": "function", "function": {
        "name": "get_weather",
        "description": "도시 날씨 조회",
        "parameters": {"type": "object",
                       "properties": {"city": {"type": "string"}},
                       "required": ["city"]}}},
    {"type": "function", "function": {
        "name": "need_umbrella",
        "description": "날씨 상태로 우산 필요 여부 판단",
        "parameters": {"type": "object",
                       "properties": {"condition": {"type": "string"}},
                       "required": ["condition"]}}},
]

tool_map = {
    "get_weather":   get_weather,
    "need_umbrella": need_umbrella,
}

messages = [{"role": "user", "content": "서울 날씨 보고 우산이 필요한지 알려줘"}]

# ── 자동 루프: Tool 호출이 없을 때까지 반복 ──
while True:
    response = chat(model="qwen2.5:7b", messages=messages, tools=tools)

    if not response.message.tool_calls:
        # Tool 호출 없음 = 최종 답변
        print(response.message.content)
        break

    messages.append(response.message)

    for tc in response.message.tool_calls:
        name   = tc.function.name
        args   = tc.function.arguments
        result = tool_map[name](**args)
        print(f"[Tool] {name}({args}) → {result}")

        messages.append({
            "role":    "tool",
            "content": json.dumps(result, ensure_ascii=False),
        })
```

```python
# 코인 유사도

from numpy import dot
from numpy.linalg import norm 

def cosine_similarity(a,b):
  return dot(a,b) / (norm(a) * norm(b))
```

```python
# 두 개의 문장이 유사한가

s1 = "파이썬은 프로그래밍 언어이다"
s2 = "Python은 개발에 사용되는 언어이다" 

el1 = ollama.embed(model="qwen3-embedding:4b", input=s1).embeddings[0]
el2 = ollama.embed(model="qwen3-embedding:4b", input=s2).embeddings[0]

print(cosine_similarity(el1, el2))
```

```python
s1 = "파이썬은 프로그래밍 언어이다"
s2 = "축구는 세계적으로 인기있는 스포츠이다" 

el1 = ollama.embed(model="qwen3-embedding:4b", input=s1).embeddings[0]
el2 = ollama.embed(model="qwen3-embedding:4b", input=s2).embeddings[0]

print(cosine_similarity(el1, el2))
```

```python
import ollama

documents = [
  "파이썬은 프로그래밍 언어이다",
  "축구는 세계적으로 인기있는 스포츠이다" ,
  "인공지능은 머신러닝을 활용한다" ,
  "서울은 대한민국의 수도이다"
]

doc_vectors = [] # 문서별 백터값

for doc in documents:
  vector = ollama.embed(model="qwen3-embedding:4b", input=doc).embeddings[0]
  doc_vectors.append(vector)

query = "AI 기술이란?"
query_vector = ollama.embed(model="qwen3-embedding:4b", input=query).embeddings[0]

scores = []
for doc, vec in zip(documents, doc_vectors):
  score = cosine_similarity(query_vector, vec) # 얼마나 유사한지 찾는다
  scores.append((doc, score))

scores.sort(key=lambda x:x[1], reverse=True) # 높은순으로 정렬

# 여기 문서로 가서, 비슷한 문장을 찾는다
# RAG
for doc, score in scores:
  print(doc, score)
```

```python
context = scores[0][0]

prompt = f"""
문서:
{context}

질문:
{query}
"""

response = ollama.chat(model="qwen2.5:7b", messages=[{"role":"user", "content":prompt}])
print(response.message.content)
```
