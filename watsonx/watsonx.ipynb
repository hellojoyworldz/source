```python
from dotenv import load_dotenv
import os
from ibm_watsonx_ai import APIClient
from ibm_watsonx_ai import Credentials
from ibm_watsonx_ai.foundation_models import ModelInference
```

```python
# .env 내용 가져오기
load_dotenv()

apiKey = os.getenv("WATSONX_API_KEY")
project_id = os.getenv("WATSONX_PROJECT_ID")
watsonx_ai_url = os.getenv("WATSONX_URL")

```

```python
# 현재 watsonx에서 제공하는 모델 알아보기

credentials = Credentials(
   url = f"{watsonx_ai_url}",
   api_key = f"{apiKey}",
)
client = APIClient(credentials)

# 모델 확인
client.foundation_models.TextModels.show()
```

```python

credentials = Credentials(
   url = f"{watsonx_ai_url}",
   api_key = f"{apiKey}",
)
client = APIClient(credentials)

model = ModelInference(
    model_id="ibm/granite-4-h-small",
    api_client=client,
    project_id=f"{project_id}",
    # 모델 모수
    params = {
        # 최대 토큰
        "max_tokens": 100
    }
)

prompt = 'How far is Paris from Bangalore?'
messages = [
    # 시스템 프롬프트
    {"role": "system", "content": "You ar a helpful assistant."}, 
     # 사용자 프롬프트
    {"role": "user", "content":prompt}
]

generated_response = model.chat(messages=messages)

print(generated_response)
print(generated_response['choices'][0]['message']['content'])
```

```python
credentials = Credentials(
   url = f"{watsonx_ai_url}",
   api_key = f"{apiKey}",
)
client = APIClient(credentials)

model = ModelInference(
    model_id="ibm/granite-4-h-small",
    api_client=client,
    project_id=f"{project_id}",
    # 모델 모수
    params = {
        # 최대 토큰
        "max_tokens": 2000
    }
)

user_prompt="""페페로니 피자 만드는 방법 설명해줘.
"""

system_propt = """
You are a cooking assistant.
- Answer only questions related to cooking, recipes, or ingredients.
- Provide clear and parcitial instructions.
- Use simple measurements and steps.
- Prefer bullet points over long paragraphs.
"""

messages = [
    {"role": "system", "content": system_propt}, 
    {"role": "user", "content":user_prompt}
]

# generated_response = model.chat(messages=messages)

# print(generated_response)
# print(generated_response['choices'][0]['message']['content'])

generated_response_stream = model.chat_stream(messages=messages)
for chunk in generated_response_stream:
    if chunk['choices']:
        print(chunk['choices'][0]['delta'].get('content',''), end="", flush=True)
```

```python
credentials = Credentials(
   url = f"{watsonx_ai_url}",
   api_key = f"{apiKey}",
)
client = APIClient(credentials)

model = ModelInference(
    model_id="ibm/granite-4-h-small",
    api_client=client,
    project_id=f"{project_id}",
    # 모델 모수
    params = {
        "max_tokens": 2000,
        "temperature":1.5
    }
)

user_prompt="무서운 이야기 만들어줄래?"
system_prompt = """너너 유명한 공포 소설 작가야."""

messages = [
    # 시스템 프롬프트
    {"role": "system", "content":system_prompt}, 
     # 사용자 프롬프트
    {"role": "user", "content":user_prompt}
]

generated_response = model.chat(messages=messages)

print(generated_response)
print(generated_response['choices'][0]['message']['content'])
```

### 토큰

```python
print(generated_response['usage'])

# completion_tokens: 모델이 생성한 응답에 사용된 토큰 수
# prompt_tokens: 모델이 입력으로 받은 내용에 사용된 토큰 수

total = 0.0000363 / 1000 * generated_response['usage']['prompt_tokens'] + 0.000265 / 1000 * generated_response['usage']['completion_tokens']
print(total * 1490)
```
