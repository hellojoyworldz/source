````python
# 주차장 프로그램
# 5대의 주차 공간 , A, B, C, D, E

# 메뉴 제공
# [1] 주차 | [2] 출차 | [3] 종료

# 1 선택 => A 자동차 들어감. 주차장 상태 ==> ['A']
# 1 선택 => B 자동차 들어감. 주차장 상태 ==> ['A','B']
# 2 선택 => B 자동차 나감. 주차장 상태 ==> ['A']

# 5대 까지만 주차 가능, 6번째 차량이 들어오려면 "주차장 만차" 표시 


top, car_name = 0, 'A'
parking_lot = []

while True:
    menu = int(input("[1] 주차 | [2] 출차 | [3] 종료"))

    if menu <= 3:
        if menu == 1:
            if top >=5:
                print("주차장 만차")
            else:
                parking_lot.append(car_name)
                print(f"{car_name} 자동차 들어감. 주차장 상태 ==> {parking_lot}")
                top += 1
                car_name = chr(ord(car_name) + 1)
        elif menu == 2:
            if top > 0:
                out_car = parking_lot.pop()
                print(f"{out_car} 자동차 나감. 주차장 상태 ==> {parking_lot}")
                top -= 1
            else:
               print("빠져나갈 자동차 없음")
        else:
            print("프로그램 종료")
            break;
    else:
        print("번호를 확인해 주세요")
````

````python
# *args : 한번에 받는다. 안들어와도 상관 없음

def add_many(*args):
    result = 0
    for i in args:
        result += i
    return result

print(add_many(15,16,28,29))
print(add_many(15,16,30))
print(add_many())
````

````python
def add_many(choice, *args):
    if choice == "mul":
        result = 1
        for i in args:
            result *= i
    elif choice == "add":
        result = 0
        for i in args:
            result += i
    return result

print(add_many("add", 1,2,3,4,5))

print(add_many("mul", 1,2,3,4,5))
````

### 클래스
- 함수, 변수를 포함한 형태
- 객체지향

````python
# 클래스 생성
class Calculator:
    def __init__(self): # 초기화
        self.result = 0

    # 메서드: 클래스에 들어있는 함수
    def add(self, num):
        self.result += num
        return self.result

    def sub(self, num):
        self.result -= num
        return self.result
````

````python
# 사용
calc = Calculator() # 객체 생성

# 함수(메소드) 호출

print(calc.add(5))
print(calc.sub(3))
````

````python
class FourCal:
    def __init__(self, first, second): # 생성자
        self.first = first
        self.second = second

    def add(self):
        return self.first + self.second

    def sub(self):
        return self.first - self.second
    
    def mul(self):
        return self.first * self.second
    
    def div(self):
        return self.first // self.second
    
    def __str__(self): # __ 언더바 2개짜리는 파이썬 내장
        return f"{self.first}, {self.second}"
    
    
````

````python
# __init__() 자동으로 호출됨
calc1 = FourCal(5,3)

# 덧셈
print(calc1.add())
print(calc1.sub())
print(calc1.mul())
print(calc1.div())
print(calc1) # <__main__.FourCal object at 0x000001F83B4291D0> 주소가 뜸. 이를 방지하기 위해 class에 __str__ 정의
````

````python
# 상속

class MoreFourCal(FourCal):
    def pow(self):
        return self.first ** self.second
    
````

````python

# super() 부모
# self 자기자신

class SafeFourCal(FourCal):
    # 오버라이딩: 자식에서 재정의
    def div(self):
        if self.second == 0:
            return 0
        else:
            return self.first / self.second
       

four1 = SafeFourCal(4,0)
four1.div()
````
