### 딕셔너리
- {key1:value1, key2:value2}
- key 값 중복 불가
- 


````python
dict1 = {"name":"Kim",'age':22}
dict2 = {1: "Kim", 2:22, 3:"서울"}
dict3 = {"arr":[1,2,3,4,5]}
````

````python
# Kim 출력
print(dict1["name"])
# print(dict1["addr"]) # 없으면 에러
print(dict1.get("name")) # 추천하는 방식
print(dict1.get("addr")) # 없으면 None
print(dict2.get(3))
print(dict2[3])


````

````python
# 삭제
del dict1["age"]
dict1
````

````python
# 추가
dict1['birth'] = "1999-09-12"
dict1
````

````python
# 추가
# rank = [15,17,19]
dict1['rank'] = [15,17,19]
dict1
````

````python

print(dict1.keys()) # key 값만 가져오기
print(dict1.values()) # value 값만 가져오기
print(dict1.items()) # key-value 쌍으로 가져오기
````

````python
# in: 포함되는가?
print('사과' in ['사과','포도','바나나'])

# key 가 포함되어있는가?
print('사과' in dict1)
print('name' in dict1) 

# not in: 포함 되어있지 않은가?
print('사과' not in ['사과','포도','바나나'])
````

### 조건문
- 조건에 맞는 상황을 판단한 후 수행하도록 처리
- ex) 현재 변수에 값이 5라면, dict1 키가 name이 포함되어 있다면
```
if 조건문: 
    조건이 참일 때 수행할 문장을 작성
```


````python
# money 10,000원 이상을 가지고 있다면 택시 타기
money = 5000
if money >= 10000:
    print("택시 타기")
````

````python
# money 10,000 미만이면 버스 타기
if money >= 10000:
    print("택시 타기")
else:
    print("버스 타기")
````

````python
# 90 이상 A, 80 이상 B, 70 이상 C, 나머지 D

jumsu = 88
if jumsu >= 90:
    print("A")
elif jumsu >= 80:
    print("B")
elif jumsu >= 70:
    print("C")
else:
    print("D")
````

````python
# 점수를 입력받아 변수에 담음 
# input()을 통해 받는 모든 것은 문자 타입
num = int(input("당신의 점수를 입력하세요"))

if num >= 90:
    grade = "A"
elif num >= 80:
    grade = "B"
elif num >= 70:
    grade = "C"
else:
    grade = "D"


# 점수가 90 이상이고 평가 등급이 A라면 합격입니다. 그 외는 불합격 입니다.
if num >= 90 and grade == "A":
    print("합격입니다")
else:
    print("불합격입니다")


````

````python
# 숫자를 입력받은 후 홀,짝 구분
# == : 같은가 / != 같지 않다

number = int(input("숫자 입력"))

if number % 2 == 0:
    print("짝수")
else:
    print("홀수")
````
